### 3 approaches for webscraping

#### Approach 1: `requests` + `BeautifulSoup`: simple but only works for HTML pages

#### Approach 2: Selenium: works well for dynamic webpages

```python
from selenium.webdriver import Chrome
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as expected_conditions

driver = Chrome()
driver.get("https://quotes.toscrape.com/login")
```

```python
driver.get("https://quotes.toscrape.com/login")

username_box = "#username"
driver.find_element(By.CSS_SELECTOR, username_box).send_keys("ftd2026_user")

password = "#password"
driver.find_element(By.CSS_SELECTOR, password).send_keys("ftd2026_pwd")

login = ".btn"
driver.find_element(By.CSS_SELECTOR, login).click()
```
